```python
#第三阶段：准备模型训练数据。
#模型先采用 自编码器（Autoencoder），这是工业声音异常检测中常见的入门方案。
#核心思路：
#用正常音频训练模型，让模型学会正常风扇声音的特征。
#测试时输入正常声音，模型应该能较好地还原。
#输入异常声音，模型通常还原得较差。
#通过比较还原误差，判断声音是否异常。
```

```python
from pathlib import Path
import numpy as np
import soundfile as sf
import librosa

# 数据集路径
normal_dir = Path(
    "/home/jovyan/work/shared_data/MIMII/0_dB/fan/id_00/normal"
)

abnormal_dir = Path(
    "/home/jovyan/work/shared_data/MIMII/0_dB/fan/id_00/abnormal"
)

# 读取音频并提取 Mel 频谱
def extract_mel(file_path):#定义函数
    audio, sr = sf.read(str(file_path))

    # 多声道转单声道
    if audio.ndim == 2:#判读音频有几个维度
        audio = np.mean(audio, axis=1)

    # 提取 Mel 频谱
    mel = librosa.feature.melspectrogram(
        y=audio,
        sr=sr,
        n_mels=128
    )

    # 转换成分贝
    mel_db = librosa.power_to_db(mel, ref=np.max)

    return mel_db.T.astype(np.float32)#转置矩阵 让每一行代表一个时间帧 转换为32浮点数
```

```python
# 获取所有正常音频文件
normal_files = sorted(normal_dir.glob("*.flac"))

print("正常音频数量：", len(normal_files))
```

```python
# 提取所有正常音频的特征
normal_features = []

for file_path in normal_files:
    features = extract_mel(file_path)
    normal_features.append(features)

print("提取完成！")
print("正常音频数量：", len(normal_features))
print("第一条音频特征形状：", normal_features[0].shape)
```

```python
# 1. 定义正常和异常音频文件夹
normal_dir = Path(
    "/home/jovyan/work/shared_data/MIMII/0_dB/fan/id_00/normal"
)

abnormal_dir = Path(
    "/home/jovyan/work/shared_data/MIMII/0_dB/fan/id_00/abnormal"
)

# 2. 获取音频文件
normal_files = sorted(normal_dir.glob("*.flac"))
abnormal_files = sorted(abnormal_dir.glob("*.flac"))

# 3. 定义 Mel 特征提取函数
def extract_mel(file_path):
    audio, sr = sf.read(str(file_path))

    if audio.ndim == 2:
        audio = np.mean(audio, axis=1)

    mel = librosa.feature.melspectrogram(
        y=audio,
        sr=sr,
        n_mels=128
    )

    mel_db = librosa.power_to_db(mel, ref=np.max)

    return mel_db.T.astype(np.float32)

# 4. 初始化特征列表
normal_features = []
abnormal_features = []

# 5. 检查
print("正常音频数量：", len(normal_files))
print("异常音频数量：", len(abnormal_files))
print("特征提取函数：", callable(extract_mel))
```
