```python
#音频预处理
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

# 1. 将8声道音频合并为单声道
mono_audio = np.mean(audio, axis=1)

# 2. 提取 Mel 频谱
mel = librosa.feature.melspectrogram(
    y=mono_audio,
    sr=sample_rate,
    n_mels=128
)

# 3. 将能量转换为分贝
mel_db = librosa.power_to_db(mel, ref=np.max)

# 4. 查看特征矩阵的形状
print("Mel频谱形状：", mel_db.shape)
```

```python
plt.figure(figsize=(12, 5))#创建画图

librosa.display.specshow(#绘制Mel频图
    mel_db,#频率数据
    sr=sample_rate,#采样率
    x_axis="time",#定义xy轴
    y_axis="mel"
)

plt.colorbar(format="%+2.0f dB")#添加颜色条
plt.title("Normal Fan Mel Spectrogram")
plt.xlabel("Time")
plt.ylabel("Mel Frequency")

plt.show()
```

```python
#提取异常音频Mel频谱
# 1. 将异常音频的8个声道合并为单声道
abnormal_mono = np.mean(abnormal_audio, axis=1)

# 2. 提取异常音频的 Mel 频谱
abnormal_mel = librosa.feature.melspectrogram(
    y=abnormal_mono,
    sr=abnormal_sample_rate,
    n_mels=128
)

# 3. 转换成分贝
abnormal_mel_db = librosa.power_to_db(
    abnormal_mel,
    ref=np.max
)

# 4. 查看特征矩阵形状
print("异常音频 Mel 频谱形状：", abnormal_mel_db.shape)
```

```python
plt.figure(figsize=(12, 5))

librosa.display.specshow(
    abnormal_mel_db,
    sr=abnormal_sample_rate,
    x_axis="time",
    y_axis="mel"
)

plt.colorbar(format="%+2.0f dB")
plt.title("Abnormal Fan Mel Spectrogram")
plt.xlabel("Time")
plt.ylabel("Mel Frequency")

plt.show()
```
